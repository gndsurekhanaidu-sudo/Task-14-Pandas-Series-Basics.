```python
import pandas as pd

# Creating a Series
marks = pd.Series(
    [85, 90, 78, 95, 88],
    index=["Ram", "Ravi", "Sita", "Priya", "John"]
)

# Display Series
print("Student Marks:")
print(marks)

# Indexing
print("\nRam Marks:", marks["Ram"])

# Filtering
print("\nMarks Greater Than 85:")
print(marks[marks > 85])

# Arithmetic Operations
print("\nAdding 5 Marks:")
print(marks + 5)

# Statistical Operations
print("\nTotal Marks:", marks.sum())
print("Average Marks:", marks.mean())
print("Maximum Marks:", marks.max())
print("Minimum Marks:", marks.min())

# Statistical Summary
print("\nStatistical Summary:")
print(marks.describe())
```
